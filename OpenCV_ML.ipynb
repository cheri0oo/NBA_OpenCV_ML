```python
# (1) load YOLO model for PLAYERS detection

from ultralytics import YOLO

# load a pretrained - models PLAYERS
model = YOLO("models/player_detector.pt")  
results = model.predict("input_videos/video_1.mp4",save=True)

print(results)
print("               ")
for box in results[0].boxes:
    print(box.xyxy, box.conf, box.cls)
```

```python
# (2) load YOLO model for BALL detection

from ultralytics import YOLO

# load a pretrained - models BALL
model = YOLO("models/ball_detector_model.pt")  
results = model.predict("input_videos/video_1.mp4",save=True)

print(results)
print("               ")
for box in results[0].boxes:
    print(box.xyxy, box.conf, box.cls)
```

```python
# (3) Tracking PLAYERS

from ultralytics import YOLO

# USE TRACK to give players ID
model = YOLO("models/player_detector.pt")  
results = model.track("input_videos/video_1.mp4",save=True)

print(results)
print("               ")
for box in results[0].boxes:
    print(box.xyxy, box.conf, box.cls)
```
